### 데이터 불러오기

```python
import pandas as pd
from scipy import stats

df = pd.read_csv('../week1/data/cookie_cats.csv')

# A, B 그룹 나누기
gate_30 = df[df['version'] == 'gate_30']
gate_40 = df[df['version'] == 'gate_40']

print(f"gate_30: {len(gate_30)}명")
print(f"gate_40: {len(gate_40)}명")
```

### STEP A. D1 잔존률 카이제곱 검정
- 잔존 여부가 True/False인 범주형 데이터이므로 chi-square 사용



```python
# 분할표 만들기
contingency_table = pd.crosstab(df['version'], df['retention_1'])
print("[ D1 잔존 분할표 ]")
print(contingency_table)

# 카이제곱 검정
chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)
print(f"\nchi2 통계량: {chi2:.4f}")
print(f"p-value: {p_value:.4f}")
print(f"자유도: {dof}")

if p_value < 0.05:
    print("결과: p < 0.05 이므로 두 그룹 간 D1 잔존률 차이는 통계적으로 유의미함")
else:
    print("결과: p >= 0.05 이므로 두 그룹 간 D1 잔존률 차이는 통계적으로 유의미하지 않음")
```

### STEP B. D7 잔존률도 동일하게 검정


```python
# 분할표 만들기
contingency_d7 = pd.crosstab(df['version'], df['retention_7'])
print("[ D7 잔존 분할표 ]")
print(contingency_d7)

# 카이제곱 검정
chi2_d7, p_value_d7, dof, _ = stats.chi2_contingency(contingency_d7)
print(f"\nchi2 통계량: {chi2_d7:.4f}")
print(f"p-value: {p_value_d7:.4f}")
print(f"자유도: {dof}")

if  p_value_d7 < 0.05:
    print("결과: p < 0.05 이므로 두 그룹 간 D7 잔존률 차이는 통계적으로 유의미함")
else:
    print("결과: p >= 0.05 이므로 두 그룹 간 D7 잔존률 차이는 통계적으로 유의미하지 않음")
```

## 분석 결론

| 지표 | gate_30 | gate_40 | p-value | 유의미? |
|------|---------|---------|---------|---------|
| D1 잔존률 | 44.8% | 44.2% | 0.0739 | ❌ |
| D7 잔존률 | 19.0% | 18.2% | 0.0001 | ✅ |

### 인사이트
- D1 잔존률은 두 그룹 간 차이가 통계적으로 유의미하지 않음
- D7 잔존률은 gate_30이 gate_40보다 유의미하게 높음 (p < 0.05)
- **결론: 게이트를 30라운드에 유지하는 것이 장기 잔존률에 유리함**
